## Non-Linear Dimensionality Reduction

Example of PCA on a dataset distribuited on a manifold, follow points, find no proper direction to distinguish

Manifold learning techniques on the other hand, do allow for this S-surface to be unwrapped so that its underlying structure is revealed. Two techniques, Locally linear embedding (LLE) and Isometric mapping (IsoMap) show the power of these approaches
NB non linear processes in general cannot be inverted 

LLE idea
1) define and compute local geometry, at each point define the local hypersurface
2) cmopute a lower dimensional embedding, keeping the geometry fix

Example of facial expressions

ISOMAP idea
Rather than using Euclidean distances between points, IsoMap actually approximates geodesic curves that lie in the embedded manifold, then computes the distances between each point in the data set along these geodesics.
A star algorith for GoogleMaps

tSNE
1) measure the similarity between points in the original high-dimensional space (perplexity and similarity matrix)
2) find a lower-dimensional (use t-student) mapping that preserves the similarities (compute distances between probability distribution)
nb: if the two distrubution are the same it doesnt work
