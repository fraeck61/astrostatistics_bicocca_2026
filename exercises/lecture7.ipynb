## Nonparametric Modeling & Histograms

### Histograms

Imagine you have some one-dimensional ("univariate") data that you would like to try to understand. Where by "understand" we mean "know the distribution in the measured space", i.e., you want to know the probability distribution function. Our constant friend is the histogram, and it's usually the first thing any of us do on new data. Simple, right? Not quite...

Let's work through some examples to see what problems we encounter and how we might overcome them.

```python
import numpy as np
from matplotlib import pyplot as plt
%matplotlib inline
%config InlineBackend.figure_format='retina'

import scipy.stats as stats
from scipy.stats import norm
from scipy.stats import uniform
import scipy.stats
from scipy import optimize
from astroML import stats as astroMLstats
```

```python
# Generate our data: a mix of several Cauchy distributions
np.random.seed(0)
N = 1000
mu_gamma_f = [(5, 1.0, 0.1),
              (7, 0.5, 0.5),
              (9, 0.1, 0.1),
              (12, 0.5, 0.2),
              (14, 1.0, 0.1)]
true_pdf = lambda x: sum([f * stats.cauchy(mu, gamma).pdf(x)
                          for (mu, gamma, f) in mu_gamma_f])
x = np.concatenate([stats.cauchy(mu, gamma).rvs(int(f * N))
                    for (mu, gamma, f) in mu_gamma_f])
np.random.shuffle(x)
x = x[x > -10]
x = x[x < 30]
```

```python
fig, axes = plt.subplots(nrows=1,ncols=3,figsize=(15,5))

axes[0].hist(x,bins=10,histtype='step', density=True) 
axes[1].hist(x,bins=20,histtype='step', density=True) 
axes[2].hist(x,bins=100,histtype='step', density=True)

plt.tight_layout()
plt.show()
```

How would you characterize this distribution? Could we reasonably think of it as a normal (Gaussian) distribution that we could characterize by some mean and standard deviation? Maybe, but even just by looking at this plot we can see that it wouldn't be a particularly good description of the data.

We find that small changes in the (hyper)parameters of the histogram function *significantly* change the PDF. That's bad, because the underlying data clearly have **not** changed.

* One of the problems with histograms is that some bins end up with little (or no) data.
* Also we're effectively "moving" data at the center of each bin. That's bad, we should never change our data.
* ...and modeling a pdf with a non-differentiable, step-like function is definitely very artificial (physical processes don't behave like that!)

We can do a bit better by making **variable-width bin sizes** that have the ***same number of objects in each bin***.

```python
bins100 = np.append(np.sort(x)[::100], np.max(x)) 
bins30 = np.append(np.sort(x)[::30], np.max(x))
print(len(bins100),len(bins30))

plt.hist(x, bins=bins100, density=True, histtype="step") 
plt.hist(x, bins=bins30, density=True, histtype="step")
plt.xlim(-5,25)

plt.show()
```

### Rule's of thumb: Scott and Freedman-Diaconis



**"Scott's rule"** suggests that the optimal bin width is

$$\Delta_b = \frac{3.5\sigma}{N^{1/3}}.$$

This makes sense only if the standard deviation $\sigma$ provides a reasonable description of the width of the distribution (e.g., if the distribution is somewhat close to Gaussian). If not, the **"Freedman-Diaconis rule"** says:

$$\Delta_b = \frac{2(q_{75} - q_{25})}{N^{1/3}} = \frac{2.7\sigma_G}{N^{1/3}}.$$

### Kernal Density Estimation (KDE)

Non parametric appoarch to try to track the underlying pdf. The data stay centered where they are (unlike in a histogram)

Just about any distribution can be used as the kernel, but the most common are a **Gaussian kernel** and an **Epanechnikov kernel**. One downside of the Gaussian kernel is that the tails are technically infinite in extent. So each point has some finite probability of being everywhere. The Epanechnikov kernel has truncated wings.

One still has the problem of deciding the width of the kernel (e.g., for the Gaussian the "mean" is fixed at the value of the point, but how wide should you make the Gaussian?). For now, we'll just play with the widths by hand to see what might work best. N.B. the widths of the kernel distribution are referred to as "bandwidth".

In practice, the **bandwidth** is almost always the key parameter of a KDE representation, the exact shape of the kernel doesn't really matter too much

```python
from sklearn.neighbors import KernelDensity
xgrid = np.linspace(x.min(),x.max(),1000)

def kde_sklearn(data, bandwidth = 1.0, kernel="linear"):
    kde_skl = KernelDensity(bandwidth = bandwidth, 
                            kernel=kernel)
    kde_skl.fit(data[:, np.newaxis])
    log_pdf = kde_skl.score_samples(xgrid[:, np.newaxis]) # sklearn returns log(density)
    return np.exp(log_pdf)
```

```python
PDFtophat = kde_sklearn(x,bandwidth=0.1,kernel="tophat") 
plt.plot(xgrid,PDFtophat)
plt.show()
```

To select **optimal bandwith** for KDE we will use **cross-validation**.

```python
PDFtophat = kde_sklearn(x,bandwidth=0.3,kernel="gaussian")
plt.plot(xgrid,PDFtophat) 

PDFtophat = kde_sklearn(x,bandwidth=0.3,kernel="epanechnikov") 
plt.plot(xgrid,PDFtophat) 

plt.show()
```

# The irreducible mass of Black Holes

### Introduciton

For a BH of mass $M$ spin $\chi \in [0, 1]$, the irreducible mass is

$$M_{\rm irr} = M \sqrt{\frac{1 + \sqrt{1 - \chi^2}}{2}}$$

For convenience, let us define

$$f(\chi) = \frac{M_{\rm irr}}{M} = \sqrt{\frac{1 + \sqrt{1 - \chi^2}}{2}}$$

Suppose you have repeated measurements of $M$ and $\chi$. This is just because we're still frequentists for now. When we will put our Bayesian hat on, these distributions will be the *posterior samples* from a single measurement.

Let's assume that

* the spin $\chi$ is uniformly distributed between 0 and 1.
* the mass $M$ is normally distributed with mean $\mu$ and standard deviation $\sigma$.
* (masses and spins are uncorrelated)

These are simplistic assumption that, however, are not too far from what we are now measuring with gravitational-wave detectors. It's a toy model, but not a totally unreasonable one.

### Task 1-5

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde, ks_2samp

np.random.seed(42) 
N_samples = 100000 
mu = 1.0           
sigma = 0.02

# sample mass and spin with montecarlo

def calcola_f(chi):
    return np.sqrt((1 + np.sqrt(1 - chi**2)) / 2)

chi_samples = np.random.uniform(0, 1, N_samples)
M_samples = np.random.normal(mu, sigma, N_samples)

f_samples = calcola_f(chi_samples)
Mirr_samples = M_samples * f_samples

fig, ax = plt.subplots(figsize=(9, 6))
```

```python
# histogram with freedman-diaconis rule for bin width
fig, ax = plt.subplots(figsize=(10, 6))
ax.hist(Mirr_samples, bins='fd', density=True, alpha=0.5, color='gray', 
        label='Istogramma $M_{irr}$')

# smooth the histogram with KDE
kde = gaussian_kde(Mirr_samples)

x_grid = np.linspace(Mirr_samples.min(), Mirr_samples.max(), 1000)
ax.plot(x_grid, kde(x_grid), color='red', lw=2, label='KDE $M_{irr}$')

ax.set_title(rf"Distribuzione di $M_{{irr}}$ ($\mu={mu}$, $\sigma={sigma}$)")
ax.set_xlabel(r"Massa irriducibile $M_{irr}$")
ax.set_ylabel("Densità di probabilità")
ax.legend()
plt.show()
```

Kolmogorov Smirnov show esiste una distnaza che dipende solo dalle CDF di ogni pdf
La distanza KS è legata a confindence e pvalue.

```python
#test KS
sigmas = np.logspace(-3, 1, 30)

ks_dist_f = []
ks_dist_M = []
p_value_f = []
p_value_M = []

# Riduciamo il campione a 10k per velocizzare i calcoli iterativi
chi_test = np.random.uniform(0, 1, 10000)
f_test = calcola_f(chi_test)

for s in sigmas:
    # Generiamo un nuovo campione di masse con la nuova sigma
    M_test = np.random.normal(mu, s, 10000)
    Mirr_test = M_test * f_test
    
    # Calcoliamo la statistica (distanza) KS tra Mirr e f
    stat_f, pval_f = ks_2samp(Mirr_test, f_test)
    ks_dist_f.append(stat_f)
    
    # Calcoliamo la statistica (distanza) KS tra Mirr e M
    stat_M, pval_M = ks_2samp(Mirr_test, M_test)
    ks_dist_M.append(stat_M)

# Plot dei risultati dei test KS
fig, ax = plt.subplots(figsize=(9, 6))
ax.plot(sigmas, ks_dist_f, 'o-', color='blue', label=r'KS Distanza: $M_{irr}$ vs $f(\chi)$')
ax.plot(sigmas, ks_dist_M, 's-', color='orange', label=r'KS Distanza: $M_{irr}$ vs $M$')

ax.set_xscale('log')
ax.set_title(r"Evoluzione della distanza KS in funzione di $\sigma$")
ax.set_xlabel(r"$\sigma$ (Scala Logaritmica)")
ax.set_ylabel(r"Statistica KS (Distanza massima tra le CDF)")
ax.grid(True, linestyle='--', alpha=0.6)
ax.legend()
plt.show()
```

### Physical Interpretation



Our KS-test analysis reveals how measurement uncertainty dictates our ability to extract physical parameters:

* **High Noise Regime ($\sigma \gg \mu$):** As the variance on the mass measurement grows, the noise completely dominates the narrow geometric bounds of the spin function $f(\chi)$. The PDF of the irreducible mass ($M_{\rm irr}$) degenerates into the PDF of the mass ($M$). In this regime, the spin information is entirely washed out by the detector's uncertainty.
* **High Accuracy Regime ($\sigma \ll \mu$):** If our detector is highly accurate, the mass $M$ is effectively constrained. The observed variations in $M_{\rm irr}$ are strictly driven by the spin distribution $f(\chi)$. Measuring $M_{\rm irr}$ allows us to directly capture the spin of the black hole. 

**Conclusion:** To extract spin information from a composite parameter like $M_{\rm irr}$, the measurement error on the mass must be strictly smaller than the intrinsic variation caused by the spin itself.
